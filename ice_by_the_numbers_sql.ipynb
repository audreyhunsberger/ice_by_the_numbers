```sql
-- The following analysis was conducted by Audrey Hunsberger. 

-- The original data table, "Detention stays," was downloaded from The Deportation Data Project (https://deportationdata.org/data/processed/ice.html#latest-data-release) on September 2, 2026 and covers data through early August 2026. It is in this notebook as workspace.default.detention_stays_latest,  but the majority of analyses are ran on the trump2_stays, which is a subset of the original table that only includes stays booked under the second Trump administration, starting on January 21, 2026, the first full day of his administration.
```

```sql
%sql --name trump2_stays
-- Reusable base filter for stays booked under second Trump administration

WITH trump2_stays AS (
  SELECT *
    FROM workspace.default.detention_stays_latest
    WHERE stay_book_in_date_time >= TIMESTAMP '2025-01-21 00:00:00'
        AND stay_book_in_date_time <  TIMESTAMP '2026-08-07 00:00:00'
)
SELECT * FROM trump2_stays
```

```sql
-- Number of people in detention, as of August 8, 2026

SELECT COUNT(DISTINCT unique_identifier) AS currently_detained
FROM workspace.default.detention_stays_latest
WHERE stay_book_out_date_time IS NULL;
```

```sql
-- Number of removals under trump 2 (includes people booked before his admin started)

SELECT COUNT(*) AS removals_trump2
FROM workspace.default.detention_stays_latest
WHERE stay_release_reason = 'Removed'
  AND stay_book_out_date_time >= TIMESTAMP '2025-01-21 00:00:00';
```

```sql
-- Number of people booked into detention under Trump 2

SELECT COUNT(DISTINCT unique_identifier) AS people_booked_trump2
FROM trump2_stays;
```

```sql
-- average detainee length of stay for detainees booked under trump 2 (completed stays only)

SELECT ROUND(AVG(stay_length_days), 1) AS avg_length_of_stay_days
FROM trump2_stays
WHERE stay_book_out_date_time IS NOT NULL;
```

```sql
-- average length of stay, by month, under trump 2

-- I limited this to April 2025-March 2026 to cut out two sources of distortion at the edges. January-March 2025 includes the partial first month of the administration (only 10 days of January) and was still ramping up operationally, so those months aren't comparable to a full month of steady-state enforcement. On the back end, I dropped the most recent months because of right-censoring: a person's length of stay only gets counted once they're released, so any month close to the data pull date is missing everyone who's still detained, which artificially deflates the average the closer you get to "now." June 2026 is the last month where at least 90% the cohort has had time to be released for the average to be reasonably reliable. The pct_released_so_far column is included for your reference.

SELECT
  DATE_TRUNC('month', stay_book_in_date_time) AS book_in_month,
  ROUND(AVG(CASE WHEN stay_book_out_date_time IS NOT NULL THEN stay_length_days END), 1) AS avg_length_of_stay_days,
  COUNT(*)                                                                   AS booked_in,
  ROUND(100.0 * COUNT(stay_book_out_date_time) / COUNT(*), 1)                AS pct_released_so_far
FROM trump2_stays
WHERE stay_book_in_date_time >= TIMESTAMP '2025-04-01 00:00:00'
  AND stay_book_in_date_time <  TIMESTAMP '2026-04-01 00:00:00'
GROUP BY 1
ORDER BY 1;
```

```sql
-- % released as of august 8, 2026, by month booked in

SELECT
  DATE_TRUNC('month', stay_book_in_date_time) AS book_in_month,
  ROUND(AVG(CASE WHEN stay_book_out_date_time IS NOT NULL THEN stay_length_days END), 1) AS avg_length_of_stay_days,
  COUNT(*)                                                                   AS booked_in,
  ROUND(100.0 * COUNT(stay_book_out_date_time) / COUNT(*), 1)                AS pct_released_so_far
FROM trump2_stays
GROUP BY 1
ORDER BY 1;
```

```sql
-- Stays by conviction type (none, nonviolent, violent)

SELECT
  COALESCE(conviction, 'Unknown')                                  AS conviction_type,
  COUNT(*)                                                          AS stays,
  ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 1)                AS pct
FROM trump2_stays
GROUP BY 1
ORDER BY stays DESC;
```

```sql
-- Stays by criminality

SELECT
  book_in_criminality,
  COUNT(*)                                                          AS stays,
  ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 1)                AS pct
FROM trump2_stays
GROUP BY 1
ORDER BY 1;
```

```sql
-- Stays by threat level

SELECT
  COALESCE(case_threat_level, 'Missing')                            AS threat_level,
  COUNT(*)                                                          AS stays,
  ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 1)                AS pct
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
GROUP BY 1
ORDER BY 1;
```

```sql
-- Most common convictions

WITH ice_total AS (
  SELECT COUNT(*) AS n_all_ice
  FROM trump2_stays
  WHERE apprehension_final_program_group = 'ICE'
)
SELECT
  msc_charge                                                        AS most_serious_conviction,
  COUNT(*)                                                          AS stays,
  ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 1)                AS pct_of_convicted,
  ROUND(100.0 * COUNT(*) / MAX(t.n_all_ice), 2)                     AS pct_of_all_ice
FROM trump2_stays
CROSS JOIN ice_total t
WHERE apprehension_final_program_group = 'ICE'
  AND book_in_criminality = '1 Convicted Criminal'
GROUP BY 1
ORDER BY stays DESC
LIMIT 20;
```

```sql
-- by conviction

SELECT
  DATE_TRUNC('MONTH', stay_book_in_date_time)                       AS month,
  COUNT(*)                                                          AS ice_book_ins,
  ROUND(100.0 * SUM(CASE WHEN conviction = 'None' THEN 1 ELSE 0 END) / COUNT(*), 1)
                                                                    AS pct_no_conviction,
  ROUND(100.0 * SUM(CASE WHEN conviction = 'Violent crime' THEN 1 ELSE 0 END) / COUNT(*), 1)
                                                                    AS pct_violent_conviction,
  ROUND(100.0 * SUM(CASE WHEN book_in_criminality = '3 Other Immigration Violator' THEN 1 ELSE 0 END) / COUNT(*), 1)
                                                                    AS pct_no_record_at_all
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
GROUP BY 1
ORDER BY 1;
```

```sql
-- Counts and conviction by age group

SELECT
  CASE
    WHEN 2026 - birth_year < 5 THEN '<4'
    WHEN 2026 - birth_year < 10 THEN '5-9'
    WHEN 2026 - birth_year < 13 THEN '10-12'
    WHEN 2026 - birth_year < 19 THEN '13-18'
    WHEN 2026 - birth_year < 30 THEN '18-29'
    WHEN 2026 - birth_year < 45 THEN '30-44'
    WHEN 2026 - birth_year < 60 THEN '45-59'
    WHEN 2026 - birth_year < 70 THEN '60-69'
    ELSE '70+'
  END                                                               AS age_band,
  COUNT(*)                                                          AS stays,
  ROUND(100.0 * SUM(CASE WHEN conviction = 'None' THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_no_conviction
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
  AND birth_year IS NOT NULL
GROUP BY 1
ORDER BY CASE age_band
  WHEN '<4' THEN 1
  WHEN '5-9' THEN 2
  WHEN '10-12' THEN 3
  WHEN '13-18' THEN 4
  WHEN '18-29' THEN 5
  WHEN '30-44' THEN 6
  WHEN '45-59' THEN 7
  WHEN '60-69' THEN 8
  WHEN '70+' THEN 9
END;
```

```sql
-- Length of stay by age group

WITH banded AS (
  SELECT
    *,
    CASE
      WHEN 2026 - birth_year < 5  THEN '<4'
      WHEN 2026 - birth_year < 10 THEN '5-9'
      WHEN 2026 - birth_year < 13 THEN '10-12'
      WHEN 2026 - birth_year < 19 THEN '13-18'
      WHEN 2026 - birth_year < 30 THEN '18-29'
      WHEN 2026 - birth_year < 45 THEN '30-44'
      WHEN 2026 - birth_year < 60 THEN '45-59'
      WHEN 2026 - birth_year < 70 THEN '60-69'
      ELSE '70+'
    END AS age_band
  FROM trump2_stays
  WHERE apprehension_final_program_group = 'ICE'
    AND birth_year IS NOT NULL
)
SELECT
  age_band,
  COUNT(*)                                                                        AS stays,
  ROUND(100.0 * SUM(CASE WHEN conviction = 'None' THEN 1 ELSE 0 END) / COUNT(*), 1)
                                                                                  AS pct_no_conviction,
  SUM(CASE WHEN stay_book_out_date_time IS NOT NULL THEN 1 ELSE 0 END)            AS completed_stays,
  ROUND(AVG(CASE WHEN stay_book_out_date_time IS NOT NULL THEN stay_length_days END), 1)
                                                                                  AS mean_days,
  ROUND(MEDIAN(CASE WHEN stay_book_out_date_time IS NOT NULL THEN stay_length_days END), 1)
                                                                                  AS median_days,
  ROUND(100.0 * SUM(CASE WHEN stay_book_out_date_time IS NOT NULL
                          AND stay_length_days >= 90 THEN 1 ELSE 0 END)
        / NULLIF(SUM(CASE WHEN stay_book_out_date_time IS NOT NULL THEN 1 ELSE 0 END), 0), 1)
                                                                                  AS pct_90_days_plus
FROM banded
GROUP BY 1
ORDER BY CASE age_band
  WHEN '<4' THEN 1 WHEN '5-9' THEN 2 WHEN '10-12' THEN 3 WHEN '13-18' THEN 4
  WHEN '18-29' THEN 5 WHEN '30-44' THEN 6 WHEN '45-59' THEN 7 WHEN '60-69' THEN 8
  WHEN '70+' THEN 9 END;
```

```sql
-- No conviction by state

SELECT
  state_first                                                       AS state,
  COUNT(*)                                                          AS ice_stays,
  ROUND(100.0 * SUM(CASE WHEN conviction = 'None' THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_no_conviction
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
GROUP BY 1
ORDER BY ice_stays DESC;
```

```sql
-- Outcomes by conviction status

SELECT
  stay_release_reason,
  COUNT(*)                                                          AS stays,
  ROUND(100.0 * SUM(CASE WHEN conviction = 'None' THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_no_conviction
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
  AND stay_book_out_date_time IS NOT NULL
GROUP BY 1
ORDER BY stays DESC
LIMIT 12;
```

```sql
-- Length of stay by crime type/conviction status

SELECT
  COALESCE(conviction, 'Unknown')                                   AS conviction_type,
  COUNT(*)                                                          AS completed_stays,
  ROUND(AVG(stay_length_days), 1)                                   AS mean_days,
  ROUND(MEDIAN(stay_length_days), 1)                                AS median_days
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
  AND stay_book_out_date_time IS NOT NULL
GROUP BY 1
ORDER BY completed_stays DESC;
```

```sql
-- 287(g) stays vs. others

SELECT
  CASE WHEN final_program IN ('287G Program', '287g Task Force') THEN '287(g) cohort'
       ELSE 'Other ICE' END                                        AS cohort,
  COUNT(*)                                                          AS stays,
  ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 1)                AS pct_of_ice_total
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
GROUP BY 1;
```

```sql
-- Conviction type, 287(g) vs. other ICE

WITH grouped AS (
  SELECT
    CASE WHEN final_program IN ('287G Program', '287g Task Force') THEN '287(g) cohort'
         ELSE 'Other ICE' END                                        AS cohort,
    COALESCE(conviction, 'Unknown')                                  AS conviction_type,
    COUNT(*)                                                          AS stays
  FROM trump2_stays
  WHERE apprehension_final_program_group = 'ICE'
  GROUP BY 1, 2
)
SELECT
  cohort,
  conviction_type,
  stays,
  ROUND(100.0 * stays / SUM(stays) OVER (PARTITION BY cohort), 1) AS pct_within_cohort
FROM grouped
ORDER BY 1, stays DESC;
```

```sql
-- convictions in 287(g) cohort

SELECT
  msc_charge,
  COUNT(*)                                                          AS stays
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
  AND book_in_criminality = '1 Convicted Criminal'
  AND final_program IN ('287G Program', '287g Task Force')
GROUP BY 1
ORDER BY stays DESC
LIMIT 10;
```

```sql
-- threat level vs violent criminal analysis

SELECT
  COALESCE(case_threat_level, 'Missing')                            AS threat_level,
  COUNT(*)                                                          AS stays,
  ROUND(100.0 * SUM(CASE WHEN conviction = 'None' THEN 1 ELSE 0 END) / COUNT(*), 1)
                                                                    AS pct_no_conviction,
  ROUND(100.0 * SUM(CASE WHEN conviction = 'Violent crime' THEN 1 ELSE 0 END) / COUNT(*), 1)
                                                                    AS pct_violent,
  ROUND(100.0 * SUM(CASE WHEN book_in_criminality = '3 Other Immigration Violator'
                         THEN 1 ELSE 0 END) / COUNT(*), 1)          AS pct_no_record_at_all
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
GROUP BY 1
ORDER BY 1;
```

```sql
-- detention days consumed, by conviction status

SELECT
  COALESCE(conviction, 'Unknown')                                   AS conviction_type,
  COUNT(*)                                                          AS completed_stays,
  ROUND(SUM(stay_length_days))                                      AS total_detention_days,
  ROUND(100.0 * SUM(stay_length_days) / SUM(SUM(stay_length_days)) OVER (), 1)
                                                                    AS pct_of_all_detention_days
FROM trump2_stays
WHERE apprehension_final_program_group = 'ICE'
  AND stay_book_out_date_time IS NOT NULL
GROUP BY 1
ORDER BY total_detention_days DESC;
```

```sql
-- month/charges

SELECT
  DATE_TRUNC('MONTH', stay_book_in_date_time)                         AS month,
  COUNT(*)                                                            AS ice_book_ins,
  ROUND(100.0 * SUM(CASE WHEN category = 'None'            THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_none,
  ROUND(100.0 * SUM(CASE WHEN category = 'Pending charges' THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_pending_charges,
  ROUND(100.0 * SUM(CASE WHEN category = 'Violent'         THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_violent,
  ROUND(100.0 * SUM(CASE WHEN category = 'Nonviolent'      THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_nonviolent,
  ROUND(100.0 * SUM(CASE WHEN category = 'Unclassified'    THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_unclassified
FROM (
  SELECT
    stay_book_in_date_time,
    CASE
      WHEN book_in_criminality = '3 Other Immigration Violator'                       THEN 'None'
      WHEN book_in_criminality = '2 Pending Criminal Charges'                         THEN 'Pending charges'
      WHEN book_in_criminality = '1 Convicted Criminal' AND conviction = 'Violent crime' THEN 'Violent'
      WHEN book_in_criminality = '1 Convicted Criminal'                               THEN 'Nonviolent'
      ELSE 'Unclassified'
    END AS category
  FROM trump2_stays
  WHERE apprehension_final_program_group = 'ICE'
) t
GROUP BY 1
ORDER BY 1;
```

```sql
-- find death outcome

SELECT stay_release_reason, COUNT(*) AS n
FROM trump2_stays
GROUP BY 1
ORDER BY n ASC;
```
